```python
import pandas as pd

df = pd.read_csv('../data/sales_raw.csv')
df.head()
```

```python
df.info()
```

```python
# Check for duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Check which rows have missing Postal Code
df[df['Postal Code'].isnull()]
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)

df.info()
```

```python
df['Postal Code'] = df['Postal Code'].fillna(0).astype(int)
```

```python
df.to_csv('../data/sales_clean.csv', index=False)
print("Saved cleaned data:", df.shape)
```

```python
# Create a Year-Month column for trend ansalysis
df['Order Month'] = df['Order Date'].dt.to_period('M')
df['Order Month'].head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

```python
import os
os.makedirs('images', exist_ok=True)
```

```python
monthly_sales_plot = df.groupby('Order Month')['Sales'].sum().reset_index()
monthly_sales_plot['Order Month'] = monthly_sales_plot['Order Month'].astype(str)

plt.plot(monthly_sales_plot['Order Month'], monthly_sales_plot['Sales'], marker='o', color='steelblue')
plt.xticks(rotation=90)
plt.title('Monthly Sales Trend (2015-2018)')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.tight_layout()
plt.savefig('images/chart1_monthly_trend.png')
plt.show()
```

```python
top_products_plot = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

plt.barh(top_products_plot.index[::-1], top_products_plot.values[::-1], color='darkorange')
plt.title('Top 10 Best-Selling Products')
plt.xlabel('Sales ($)')
plt.tight_layout()
plt.savefig('images/chart2_top_products.png')
plt.show()
```

```python
category_plot = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

plt.bar(category_plot.index, category_plot.values, color=['#4C72B0', '#DD8452', '#55A868'])
plt.title('Total Sales by Category')
plt.ylabel('Sales ($)')
plt.tight_layout()
plt.savefig('images/chart3_category_sales.png')
plt.show()
```

```python
region_category = df.pivot_table(index='Region', columns='Category', values='Sales', aggfunc='sum')

sns.heatmap(region_category, annot=True, fmt='.0f', cmap='YlOrRd')
plt.title('Sales Heatmap: Region vs Category')
plt.tight_layout()
plt.savefig('images/chart4_region_category_heatmap.png')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, r2_score
import numpy as np

# Use the monthly_sales_plot data (Order Month, Sales) from Chart 1
model_df = monthly_sales_plot.copy()
model_df['Month Number'] = range(1, len(model_df) + 1)  # 1, 2, 3... 48 for 48 months

model_df.head()
```

```python
X = model_df[['Month Number']]
y = model_df['Sales']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

print("Train size:", len(X_train), "Test size:", len(X_test))
```

```python
model = LinearRegression()
model.fit(X_train, y_train)

print("Slope (avg sales growth per month):", model.coef_[0].round(2))
print("Intercept:", model.intercept_.round(2))
```

```python
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error:", round(mae, 2))
print("R² Score:", round(r2, 3))
```

```python
model_df['Calendar Month'] = model_df['Order Month'].apply(lambda x: int(x.split('-')[1]))

# One-hot encode calendar month so the model treats each month as its own category
month_dummies = pd.get_dummies(model_df['Calendar Month'], prefix='month', drop_first=True)
model_df2 = pd.concat([model_df, month_dummies], axis=1)

feature_cols = ['Month Number'] + list(month_dummies.columns)
X2 = model_df2[feature_cols]
y2 = model_df2['Sales']

X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.2, shuffle=False)

model2 = LinearRegression()
model2.fit(X2_train, y2_train)

y2_pred = model2.predict(X2_test)
print("MAE:", round(mean_absolute_error(y2_test, y2_pred), 2))
print("R² Score:", round(r2_score(y2_test, y2_pred), 3))
```

```python
plt.figure(figsize=(10, 5))
plt.plot(range(len(y2_test)), y2_test.values, marker='o', label='Actual', color='steelblue')
plt.plot(range(len(y2_test)), y2_pred, marker='o', label='Predicted', color='darkorange')
plt.title('Actual vs Predicted Sales (Test Set)')
plt.xlabel('Test Month Index')
plt.ylabel('Sales ($)')
plt.legend()
plt.tight_layout()
plt.savefig('images/chart5_actual_vs_predicted.png')
plt.show()
```
